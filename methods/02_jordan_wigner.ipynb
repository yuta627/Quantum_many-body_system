# Jordan-Wigner変換

1次元のスピン1/2の系を、フェルミオンの問題に書き換える方法。書き換えた結果がフェルミオンの2次形式（自由フェルミオン）になる模型では、$2^n\times2^n$ではなく$2n\times2n$の行列の対角化だけで厳密に解ける。

各模型への適用（どの項がどうなるか、行列$A, B$の具体形、手計算）は、`models/<模型>/02_jordan_wigner.ipynb` にまとめてある。

## 1. フェルミオンの生成・消滅演算子を作る

各サイトの2つの状態を「フェルミオンがいない/いる」とみなす。どのPauli演算子の固有状態を空/占有にとるかは模型に合わせて選ぶ（たとえば横磁場イジングモデルでは$X$の固有状態を使う）。以下では一般論として、$Z$の固有状態を使い

- $\ket{0}$ : フェルミオンがいない（$n_j=0$）
- $\ket{1}$ : フェルミオンがいる（$n_j=1$）

とする。別の軸を使う場合は、以下の式で$X, Y, Z$の名前を入れ替えればよい。

> #### フェルミオンとは
>

ここで、あるサイトの「フェルミオンを1個消す/作る」演算子を考える。まず思いつくのは

- 消滅演算子：$a = \ket{0}\bra{1} = \frac{X + iY}{2}$
- 生成演算子：$a^{\dagger} = \ket{1}\bra{0} = \frac{X - iY}{2}$

のような演算子である。これはそれぞれの状態に作用させることで確かめられる。これは$a^2=(a^{\dagger})^2=0$よりパウリの排他原理も満たす。しかし、別のサイトへの演算子同士の関係を考えた際に$[a_i, a_j]=0$と可換であり、フェルミオンの反交換な性質と反する。

> フェルミオンは演算子をかける順序を逆にすると「−」が出る、すなわち反交換関係にある。
>
> $\Rightarrow c_i c_j = -c_j c_i \iff \{c_i, c_j\}=0$

そこで、左側のサイトのパリティ$Z_k = 1 - 2n_k = (-1)^{n_k}$を「ひも（string）」として掛けた演算子

\begin{equation}
c_j = \left(\prod_{k<j} Z_k\right) a_j
\end{equation}

を考える。$i<j$のとき$c_j$のひもには$Z_i$が含まれ、$Z_i$は$a_i$と反交換するので、

\begin{equation}
\{c_i, c_j^\dagger\} = \delta_{ij},\quad \{c_i, c_j\} = 0
\end{equation}

というフェルミオンの反交換関係が成り立つ。これが**Jordan-Wigner変換**である。

## 2. 変換の規則：どの項が何になるか

ひもを$S_j = \prod_{k<j}Z_k$と書くと、各Pauli演算子は

\begin{equation}
Z_j = 1 - 2c_j^\dagger c_j,\quad X_j = S_j\,(c_j + c_j^\dagger),\quad Y_j = S_j\,i(c_j^\dagger - c_j)
\end{equation}

となる。$Z_j$は1サイトで閉じるが、$X_j, Y_j$には長いひもがつく。

隣り合うサイトの積では、$Z_k^2 = 1$よりひもがほぼ打ち消し合い、$Z_j$が1個だけ残る。$c_jc_j^\dagger c_j = c_j,\ c_j^\dagger c_j^\dagger c_j = 0$を使うと$(c_j + c_j^\dagger)Z_j = c_j^\dagger - c_j$なので、たとえば

\begin{equation}
X_jX_{j+1} = (c_j + c_j^\dagger)\,Z_j\,(c_{j+1} + c_{j+1}^\dagger) = (c_j^\dagger - c_j)(c_{j+1} + c_{j+1}^\dagger)
\end{equation}

となる。同様にして、次の表が得られる。

| スピンの項 | フェルミオンの項 | 次数 |
|---|---|---|
| $Z_j$ | $1 - 2c_j^\dagger c_j$ | 2次（化学ポテンシャル） |
| $X_jX_{j+1}$ | $(c_j^\dagger - c_j)(c_{j+1} + c_{j+1}^\dagger)$ | 2次 |
| $Y_jY_{j+1}$ | $-(c_j^\dagger + c_j)(c_{j+1}^\dagger - c_{j+1})$ | 2次 |
| $X_jX_{j+1} + Y_jY_{j+1}$ | $2(c_j^\dagger c_{j+1} + c_{j+1}^\dagger c_j)$ | 2次（ホッピング） |
| $X_jX_{j+1} - Y_jY_{j+1}$ | $2(c_j^\dagger c_{j+1}^\dagger + c_{j+1}c_j)$ | 2次（ペアの生成・消滅） |
| $Z_jZ_{j+1}$ | $(1-2n_j)(1-2n_{j+1})$ | **4次**（相互作用） |
| $X_j$、$Y_j$（単独） | $S_j(c_j + c_j^\dagger)$など | 長いひもが残る |
| $X_iX_j$（$j>i+1$） | 間のひも$Z_{i+1}\cdots Z_{j-1}$が残る | 2次でない |

まとめると、**ひもの軸（$Z$）の1サイトの項と、それと垂直な軸（$X, Y$）の最近接の積だけからなるハミルトニアン**は2次形式になる。

> **周期境界条件の場合**
>
> 境界の項$X_{n-1}X_0$では左右のひもが打ち消し合わず、全パリティ$P = \prod_k Z_k = (-1)^{N}$（$N$は全粒子数）が残る。
>
> \begin{equation}
> X_{n-1}X_0 = -P\,(c_{n-1}^\dagger - c_{n-1})(c_0 + c_0^\dagger)
> \end{equation}
>
> $P$はハミルトニアンと交換する（2次形式は粒子をペアでしか増減させない）ので、$P=\pm1$の部分空間ごとに、フェルミオンの境界条件を反周期/周期として解けばよい。

## 3. 2次形式の対角化（BdG形式）

ハミルトニアンが

\begin{equation}
H = \sum_{ij} c_i^\dagger A_{ij} c_j + \frac{1}{2}\sum_{ij}\left(c_i^\dagger B_{ij} c_j^\dagger + \mathrm{h.c.}\right) + \mathrm{const}
\end{equation}

と書けたとする（$A$はエルミート行列、$B$は反対称行列）。生成・消滅演算子を並べた$2n$成分のベクトル（Nambu表示）

\begin{equation}
\Psi = (c_1, \cdots, c_n, c_1^\dagger, \cdots, c_n^\dagger)^T
\end{equation}

を導入し、$\sum_{ij} c_i^\dagger A_{ij} c_j = \frac{1}{2}\left(c^\dagger A c - c^T A^T c^\dagger\right) + \frac{1}{2}\mathrm{tr}A$と書き直すと、

\begin{equation}
H = \frac{1}{2}\Psi^\dagger H_{\mathrm{BdG}}\Psi + \frac{1}{2}\mathrm{tr}A + \mathrm{const},\quad
H_{\mathrm{BdG}} = \begin{pmatrix} A & B \\ -B^* & -A^* \end{pmatrix}
\end{equation}

となる（BdGはBogoliubov-de Gennesの略）。$A, B$が実数なら$H_{\mathrm{BdG}} = \begin{pmatrix} A & B \\ -B & -A \end{pmatrix}$である。

$H_{\mathrm{BdG}}$の固有値は必ず$\pm\varepsilon_k\ (\varepsilon_k\ge 0,\ k=1,\cdots,n)$の対で現れる（粒子と正孔を入れ替える操作で符号が反転するため）。$H_{\mathrm{BdG}}$を対角化するユニタリ行列で$\Psi$を変換した新しいフェルミオン$\eta_k$（**Bogoliubov変換**）を使うと

\begin{equation}
H = \sum_{k=1}^{n}\varepsilon_k\left(\eta_k^\dagger\eta_k - \frac{1}{2}\right) + \frac{1}{2}\mathrm{tr}A + \mathrm{const}
\end{equation}

となる。$\varepsilon_k\ge0$なので、どの$\eta_k$も占有しない状態が基底状態であり

\begin{equation}
E_0 = -\frac{1}{2}\sum_{k=1}^{n}\varepsilon_k + \frac{1}{2}\mathrm{tr}A + \mathrm{const}
\end{equation}

である。励起状態は$E_0$に$\varepsilon_k$をいくつか足したものになるので、$2^n$個の多体の固有値が$n$個の1粒子エネルギーの組み合わせで表される。

## 4. 時間発展と期待値

Heisenberg描像$\Psi(t) = e^{iHt}\Psi e^{-iHt}$で考える。交換関係を計算すると$[H, \Psi] = -H_{\mathrm{BdG}}\Psi$となるので、

\begin{equation}
\frac{d}{dt}\Psi(t) = i[H, \Psi(t)] = -iH_{\mathrm{BdG}}\Psi(t)
\Rightarrow \Psi(t) = U(t)\Psi,\quad U(t) = e^{-iH_{\mathrm{BdG}}t}
\end{equation}

つまり演算子の時間発展は$2n\times2n$行列$U(t)$を掛けるだけで求まる。$U(t)$を$n\times n$のブロックに分けると

\begin{equation}
U(t) = \begin{pmatrix} U_{11} & U_{12} \\ U_{21} & U_{22} \end{pmatrix}
\Rightarrow
c_j(t) = \sum_k\left[(U_{11})_{jk}c_k + (U_{12})_{jk}c_k^\dagger\right]
\end{equation}

である。粒子数$n_j(t) = c_j^\dagger(t)c_j(t)$は初期の演算子の2次式なので、その期待値は初期状態の2点相関

\begin{equation}
F_{ba} = \bra{\psi_0}\Psi_b^\dagger\Psi_a\ket{\psi_0}\quad(2n\times2n\text{行列})
\end{equation}

だけで決まり、$\langle n_j(t)\rangle = \sum_{a,b}U_{jb}^*(t)\,F_{ba}\,U_{ja}(t)$となる。

特に初期状態がフェルミオンの真空$\ket{0}$（全サイト空）のときは、$c_k\ket{0}=0$より$c_kc_k^\dagger$の項だけが残り

\begin{equation}
\bra{0}c_j^\dagger(t)c_j(t)\ket{0} = \sum_k\left|(U_{12}(t))_{jk}\right|^2
\end{equation}

となる。$U_{12}$は「真空から粒子のペアがどれだけ生まれたか」を表している。

一方、フェルミオンで書いたときに長いひもが残る観測量（ひもの軸と垂直な1サイトの演算子$X_j$など）の期待値は、多数のフェルミオン演算子の積になるため、この方法では簡単には求まらない。

## 5. 使えない場合

2節の表で2次形式にならない項（$Z_jZ_{j+1}$のような4次の項、1サイトだけの$X_j$、離れたサイトの積）を含む模型では、相互作用のあるフェルミオンの問題になり、この方法では厳密に解けない。たとえば
- XXZ模型・ハイゼンベルク模型（$Z_jZ_{j+1}$の項を含む）：Bethe仮説など別の方法で解く
- 横磁場イジングモデルに縦磁場を加えたもの、2次元の模型：厳密対角化、テンソルネットワーク、量子回路などを使う
