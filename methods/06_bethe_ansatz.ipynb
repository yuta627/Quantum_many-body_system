# Bethe仮説（Bethe ansatz）

1次元の**可積分**な量子多体系で、固有状態を「平面波の重ね合わせ」の形で厳密に書き下す方法。H. Bethe がハイゼンベルク模型を解くために導入した（H. Bethe, Z. Phys. 71, 205 (1931)）。相互作用があっても、粒子どうしの散乱が「2粒子の散乱の積」に分解できる系では、$M$粒子の固有値問題が$M$本の連立方程式（**Bethe方程式**）に帰着する。

Jordan-Wigner変換（[02_jordan_wigner.ipynb](02_jordan_wigner.ipynb)）で解けるのは、相互作用のない自由フェルミオンに帰着する模型だけだった。Bethe仮説は、**相互作用がある**のに厳密に解ける模型を扱える。そのかわり、使えるのは特別な構造（可積分性）をもつ模型に限られる。

各模型への適用（1粒子の分散、散乱行列、Bethe方程式の具体形、手計算）は、`models/<模型>/02_jordan_wigner.ipynb` にまとめてある（Jordan-Wigner変換では解けない模型で、その代わりの厳密解として扱う）。

## 1. 考え方の流れ

厳密対角化では$2^n\times2^n$の行列を扱うが、Bethe仮説では次の順に問題を小さくしていく。

1. **保存する粒子数で分ける**：粒子数$M$ごとの部分空間で考える。
2. **1粒子**：並進対称性から、固有状態は平面波$e^{ikx}$で、エネルギーは分散関係$\varepsilon(k)$で決まる。
3. **2粒子**：粒子が離れているときは2つの平面波の積で書ける。粒子がぶつかるときの条件から、2つの平面波の振幅の比（**散乱行列**$S$）が決まる。
4. **$M$粒子**：2粒子の散乱行列の積で全体の振幅が決まる、と仮定する（**Bethe仮説**）。この仮定が矛盾なく成り立つのが可積分な模型である。
5. **境界条件**：周期境界条件を課すと、運動量$k_j$が満たすべき連立方程式（**Bethe方程式**）が得られる。
6. **解く**：$M$本の非線形方程式を数値的に解けば、エネルギー$E = \sum_j\varepsilon(k_j)$が求まる。$n\to\infty$では積分方程式になり、解析的に解けることも多い。

計算量は、$\binom{n}{M}$次元の行列の対角化から、$M$変数の連立方程式の求根に下がる。基底状態だけなら、$n\sim1000$でも簡単に計算できる。

## 2. 前提：保存する粒子数と参照状態

- **保存量**：ハミルトニアン$H$と交換する「粒子数」$\hat{N}$があるとする（$[H, \hat{N}] = 0$）。スピン系なら、Jordan-Wigner変換の一般論と同じく$Z$を粒子数の軸にとり、$\ket{0}$を空、$\ket{1}$を占有として$\hat{N} = \sum_j\frac{1 - Z_j}{2}$とする（$\sum_j Z_j$の保存）。このとき$H$は粒子数$M = 0, 1, \dots, n$のブロックに分かれる（[01_exact_diag.ipynb](01_exact_diag.ipynb)の4節）。
- **参照状態**：粒子が1個もない状態$\ket{\Omega} = \ket{0\cdots0}$は$H$の固有状態である（$M=0$のブロックは1次元）。Bethe仮説では、この状態の上に粒子を$M$個のせた状態を考える。スピン系では、参照状態の上の粒子（1個だけ反転したスピン）を**マグノン**と呼ぶ。
- **並進対称性**：サイトが一様に並び、$H$が並進$x\to x+1$で変わらないとする。境界条件はまず無限に長い鎖で考え、最後に周期境界条件を課す。
- **相互作用の到達距離**：粒子は隣のサイトへ飛び移り（ホッピング）、相互作用は短距離（最近接など）とする。

$M$粒子の状態は、粒子の座標$x_1 < x_2 < \cdots < x_M$についての波動関数$\psi(x_1, \dots, x_M)$で
\begin{equation}
\ket{\psi} = \sum_{x_1 < \cdots < x_M}\psi(x_1, \dots, x_M)\ket{x_1, \dots, x_M}
\end{equation}
と書ける（$\ket{x_1, \dots, x_M}$はサイト$x_1, \dots, x_M$だけが占有された計算基底の状態）。同じサイトに2個は入れないので$x_j$はすべて異なる。

## 3. 1粒子：平面波と分散関係

$M=1$では、$H$は1個の粒子が格子の上を動く問題になる。並進対称性から、固有状態は平面波
\begin{equation}
\psi(x) = e^{ikx},\quad H\ket{\psi} = \left(E_\Omega + \varepsilon(k)\right)\ket{\psi}
\end{equation}
である。$E_\Omega$は参照状態のエネルギー、$\varepsilon(k)$は粒子1個あたりのエネルギー（**分散関係**）である。たとえば隣のサイトへの振幅$t$のホッピング$t\,(\ket{x+1}\bra{x} + \ket{x}\bra{x+1})$があれば、$\varepsilon(k)$には$2t\cos k$が含まれる。粒子を置いたことで参照状態から変わる相互作用のエネルギーも、定数として$\varepsilon(k)$に入る。

周期境界条件$\psi(x+n) = \psi(x)$では$e^{ikn} = 1$、つまり$k = 2\pi I/n$（$I$は整数）に量子化される。$M=1$では粒子どうしの散乱がないので、これで終わりである。

## 4. 2粒子：散乱行列

$M=2$で、無限に長い鎖を考える。

- **粒子が離れているとき**（$x_2 > x_1 + 1$など、相互作用の届かない配置）

  2つの粒子は独立に動くので、平面波の積$e^{i(k_1x_1 + k_2x_2)}$はエネルギー$\varepsilon(k_1) + \varepsilon(k_2)$のシュレディンガー方程式を満たす。運動量を入れ替えた$e^{i(k_2x_1 + k_1x_2)}$も同じエネルギーをもつので、その重ね合わせ
  \begin{equation}
  \psi(x_1, x_2) = A_{12}\,e^{i(k_1x_1 + k_2x_2)} + A_{21}\,e^{i(k_2x_1 + k_1x_2)}\quad(x_1 < x_2)
  \end{equation}
  も、粒子が離れている配置では同じ方程式を満たす。$A_{12}$は「左の粒子が$k_1$、右の粒子が$k_2$」の振幅、$A_{21}$はその逆の振幅である。

- **粒子がぶつかるとき**（$x_2 = x_1 + 1$など）

  隣り合う配置では、ホッピングの行き先が埋まっていたり、相互作用が働いたりするので、離れているときの方程式とは形が変わる。2つの方程式の差（**接触条件**）を上の$\psi$が満たすように、振幅の比が決まる。
  \begin{equation}
  S(k_1, k_2) \equiv \frac{A_{21}}{A_{12}}
  \end{equation}
  これが**2粒子の散乱行列**（1次元なので1つの複素数）である。「$k_1$の粒子が左、$k_2$の粒子が右」の状態がぶつかって、「$k_2$が左、$k_1$が右」の状態になる（運動量を交換して通り抜ける）ときの振幅を表す。

1次元では、エネルギー$\varepsilon(k_1)+\varepsilon(k_2)$と全運動量$k_1+k_2$の保存から、2粒子の弾性散乱では運動量の組$\{k_1, k_2\}$そのものが変わらない（入れ替わるだけ）。このため、2粒子の散乱は1つの位相$S$だけで表される。

$S$の主な性質は次の通りである。

- 実数の$k_1, k_2$では$|S(k_1, k_2)| = 1$（ユニタリ性。確率が保存するため）
- $S(k_1, k_2)\,S(k_2, k_1) = 1$（定義から$A_{21}/A_{12}$と$A_{12}/A_{21}$）
- 相互作用がない（自由フェルミオンと同じ）ときは$S = -1$（反対称な波動関数）

> **例1：最近接ホッピングと最近接相互作用をもつ格子の粒子**
>
> 1つのサイトに1個までしか入れない粒子が、振幅$t$で隣のサイトへ飛び移り、隣り合うと相互作用$V$を感じるとする。接触条件は、$x_2 = x_1+1$のときに離れた配置の方程式で現れる「存在しない配置」$\psi(x, x)$, $\psi(x+1, x+1)$の寄与が相互作用の項と打ち消し合うこと
> \begin{equation}
> t\left[\psi(x, x) + \psi(x+1, x+1)\right] = V\psi(x, x+1)
> \end{equation}
> である。上の$\psi$を代入して$A_{21}/A_{12}$を求めると
> \begin{equation}
> S(k_1, k_2) = -\frac{1 + e^{i(k_1+k_2)} - (V/t)\,e^{ik_2}}{1 + e^{i(k_1+k_2)} - (V/t)\,e^{ik_1}}
> \end{equation}
> となる。$V = 0$なら$S = -1$である。スピン系ではXXZ模型がこの形になる（`models/02_xxz/02_jordan_wigner.ipynb`）。
>
> **例2：連続空間のデルタ関数相互作用（Lieb-Liniger模型）**
>
> 1次元の連続空間で、ボース粒子が$2c\,\delta(x_1 - x_2)$で相互作用する場合、$\varepsilon(k) = k^2$、$S(k_1, k_2) = \dfrac{k_1 - k_2 - ic}{k_1 - k_2 + ic}$となる（E. H. Lieb and W. Liniger, Phys. Rev. 130, 1605 (1963)）。$c\to\infty$で$S\to-1$（自由フェルミオンと同じ）、$c=0$で$S=1$（自由ボース粒子）になる。$S$が運動量の差だけで決まることに注意する。

## 5. $M$粒子：Bethe仮説と因子化

$M$粒子でも、粒子がすべて離れている配置では$M$個の平面波の積が固有状態の方程式を満たす。運動量の割り当て方は$M!$通りあるので、それらの重ね合わせ
\begin{equation}
\psi(x_1, \dots, x_M) = \sum_{P\in S_M} A_P\,\exp\left(i\sum_{j=1}^{M}k_{P_j}x_j\right)\quad(x_1 < \cdots < x_M)
\end{equation}
を考える（$P$は$1, \dots, M$の並べ替えで、左から$j$番目の粒子に運動量$k_{P_j}$を割り当てる）。エネルギーは
\begin{equation}
E = E_\Omega + \sum_{j=1}^{M}\varepsilon(k_j)
\end{equation}
と、1粒子のエネルギーの和になる。

**Bethe仮説**は、隣り合う2つの粒子がぶつかるときの条件が、その2粒子だけの散乱行列で満たされると仮定することである。すなわち、並べ替え$P$で隣り合う位置にある運動量$k_a$（左）と$k_b$（右）を入れ替えた並べ替えを$P'$とすると
\begin{equation}
A_{P'} = S(k_a, k_b)\,A_P
\end{equation}
とする。どの並べ替えも隣どうしの入れ替えを繰り返して作れるので、すべての$A_P$が2粒子の$S$の積で決まる。

- **因子化と可積分性**

  $M\ge3$では、たとえば3粒子の並び$(1,2,3)$から$(3,2,1)$へは、入れ替えの順序が2通りある。
  \begin{equation}
  S_{12}S_{13}S_{23} = S_{23}S_{13}S_{12}\qquad(S_{ab} = S(k_a, k_b))
  \end{equation}
  1次元で$S$が数（1成分）なら左辺と右辺は自動的に等しいが、本当に問題になるのは、3粒子が同時に近づく配置（$x_1, x_1+1, x_1+2$など）での条件が、2粒子の接触条件だけで満たされるかどうかである。満たされる模型では、多体の散乱が2体の散乱の積に分解（**因子化**）し、散乱によって新しい運動量が生まれない（**回折がない**）。粒子に内部自由度があって$S$が行列になる場合は、上の式は行列の等式（**Yang-Baxter方程式**）になり、非自明な条件になる。

  これが成り立つのは、無限個の保存量をもつ特別な模型（**可積分系**）だけである。たとえば次近接の相互作用を加えると一般に成り立たなくなり、Bethe仮説は使えない。

## 6. 周期境界条件：Bethe方程式

長さ$n$の周期境界条件では、左端の粒子を1周させて右端に持ってきても波動関数が変わらない。
\begin{equation}
\psi(x_1, x_2, \dots, x_M) = \psi(x_2, \dots, x_M, x_1 + n)
\end{equation}

- **2粒子の場合**

  右辺に4節の$\psi$を代入すると
  \begin{equation}
  \psi(x_2, x_1+n) = A_{12}\,e^{i(k_1x_2 + k_2x_1)}e^{ik_2n} + A_{21}\,e^{i(k_2x_2 + k_1x_1)}e^{ik_1n}
  \end{equation}
  である。左辺$\psi(x_1, x_2)$と$e^{i(k_1x_1+k_2x_2)}$の係数を比べると$A_{12} = A_{21}e^{ik_1n}$、つまり
  \begin{equation}
  e^{ik_1n} = \frac{A_{12}}{A_{21}} = S(k_2, k_1)
  \end{equation}
  を得る（$e^{i(k_2x_1+k_1x_2)}$の係数からは$e^{ik_2n} = S(k_1, k_2)$）。

- **$M$粒子の場合**

  同様に、$k_j$を左端から右端へ移すには、他の$M-1$個の運動量$k_l$を1回ずつ追い越す必要があり、そのたびに$S$が1つかかる。結果は
  \begin{equation}
  e^{ik_jn} = \prod_{l\neq j}S(k_l, k_j)\qquad(j = 1, \dots, M)
  \end{equation}
  となる。これが**Bethe方程式**である。

左辺は粒子$j$が自由に1周したときの位相、右辺は1周する間に他の粒子全員と1回ずつ散乱したときの位相である。Bethe方程式は「1周したときの位相の合計が1に戻る」という**量子化条件**であり、3節の$e^{ikn}=1$に散乱の位相を加えたものになっている。$S = -1$（自由フェルミオン）なら$e^{ik_jn} = (-1)^{M-1}$となり、$k_j$は互いに独立に量子化される。

解$\{k_1, \dots, k_M\}$（すべて異なる）が1つ求まるごとに、固有状態が1つ決まり、エネルギーは$E = E_\Omega + \sum_j\varepsilon(k_j)$、全運動量は$K = \sum_j k_j$である。

## 7. 対数形と量子数

実数の$k$では$|S| = 1$なので、散乱位相$\Theta$を$S(k_l, k_j) = e^{-i\Theta(k_j, k_l)}$で定義できる（$\Theta(k_j, k_l) = -\Theta(k_l, k_j)$）。Bethe方程式の対数をとると
\begin{equation}
n\,k_j = 2\pi J_j - \sum_{l\neq j}\Theta(k_j, k_l)
\end{equation}
となる。$J_j$は対数の分枝から来る整数で、$\Theta$の定義に定数（$\pi$など）を含めるかどうかで半整数になることもある。

- **量子数による状態の指定**：$J_j$を1組選ぶと、そこから$k_j$が（多くの場合ただ1つ）決まる。自由フェルミオンの$k_j = 2\pi J_j/n$と同じように、互いに異なる量子数$J_j$の組が固有状態のラベルになる。散乱の効果は、$k_j$を自由な値からずらす「位相のずれ」として入る。
- **基底状態**：エネルギーが最小になるのは、多くの模型で量子数を原点のまわりに隙間なく並べたとき（フェルミの海に相当）である。
  \begin{equation}
  J_j \to -\frac{M-1}{2}, -\frac{M-3}{2}, \dots, \frac{M-1}{2}
  \end{equation}
  （どの点を原点とするかや、整数か半整数かは模型と$\Theta$の定義による）。
- **励起状態**：量子数の並びに隙間（**ホール**）を作ったり、端の外に量子数を移したりすると励起状態になる。

## 8. ラピディティと数値解法

多くの可積分模型では、運動量$k$の代わりに**ラピディティ**$\lambda$という変数を使うと、散乱位相が**差**$\lambda_j - \lambda_l$だけの関数になる。このとき運動量を$k(\lambda)$、散乱位相を$\theta_2(\lambda_j - \lambda_l)$と書き、模型ごとの定数を量子数に含めて$I_j$と書くと、Bethe方程式は
\begin{equation}
n\,\theta_1(\lambda_j) = 2\pi I_j + \sum_{l\neq j}\theta_2(\lambda_j - \lambda_l)\qquad(j = 1, \dots, M)
\end{equation}
という形になる（$\theta_1$は$k(\lambda)$に模型ごとの定数や符号をつけたもの）。エネルギーは1粒子エネルギー$\varepsilon(\lambda)$の和
\begin{equation}
E = E_\Omega + \sum_{j=1}^{M}\varepsilon(\lambda_j)
\end{equation}
である。例2（Lieb-Liniger模型）では$k$そのものがラピディティになっている。

この連立方程式は、次のように数値的に解く。

1. **量子数を選ぶ**：基底状態なら$I_j = -\frac{M-1}{2}, \dots, \frac{M-1}{2}$。
2. **初期値を作る**：散乱を無視した$\lambda_j^{(0)} = \theta_1^{-1}(2\pi I_j/n)$や、9節の熱力学極限での分布から作った値を使う。
3. **反復またはニュートン法**：$\lambda_j \leftarrow \theta_1^{-1}\left(\frac{1}{n}\left[2\pi I_j + \sum_l\theta_2(\lambda_j - \lambda_l)\right]\right)$を収束するまで繰り返す（必要なら古い値と混ぜて安定させる）か、残差$n\theta_1(\lambda_j) - 2\pi I_j - \sum_l\theta_2(\lambda_j - \lambda_l)$をニュートン法系の解法で0にする。
4. **確かめる**：残差が十分小さいこと、小さい$n$で厳密対角化と一致することを確認する。

未知数は$M$個、1回の残差の計算は$O(M^2)$なので、$n = 1000$（$M = 500$）程度でも数秒で解ける。厳密対角化の行列の次元$\binom{1000}{500}\approx10^{299}$と比べると、圧倒的に小さい。

## 9. 熱力学極限：積分方程式

$n\to\infty$（$M/n$は一定）では、基底状態のラピディティは実軸上に密に並ぶので、和を積分に置き換えられる。**計数関数**
\begin{equation}
z(\lambda) = \frac{1}{2\pi}\left[\theta_1(\lambda) - \frac{1}{n}\sum_{l}\theta_2(\lambda - \lambda_l)\right]
\end{equation}
を定義すると、Bethe方程式は$z(\lambda_j) = I_j/n$と書ける。量子数$I_j$は間隔1で並ぶので、$\rho(\lambda) = dz/d\lambda$は「単位長さ・単位ラピディティあたりの解の数」（**分布関数**）である。和$\frac{1}{n}\sum_l f(\lambda_l)$を$\int f(\mu)\rho(\mu)\,d\mu$で置き換えて微分すると
\begin{equation}
\rho(\lambda) = a_1(\lambda) - \int_{-\Lambda}^{\Lambda}a_2(\lambda - \mu)\,\rho(\mu)\,d\mu,\qquad a_m(\lambda) = \frac{1}{2\pi}\frac{d\theta_m}{d\lambda}
\end{equation}
という線形の積分方程式になる（Lieb-Liniger模型ではLieb方程式と呼ばれる）。$\pm\Lambda$は解の分布の端で、粒子数の条件$\int_{-\Lambda}^{\Lambda}\rho(\lambda)\,d\lambda = M/n$で決まる。1サイトあたりのエネルギーは
\begin{equation}
e_0 = \lim_{n\to\infty}\frac{E_0}{n} = \frac{E_\Omega}{n} + \int_{-\Lambda}^{\Lambda}\varepsilon(\lambda)\,\rho(\lambda)\,d\lambda
\end{equation}
である。

分布が実軸全体に広がる（$\Lambda = \infty$）場合は、積分が畳み込みになるので、フーリエ変換$\hat{f}(\omega) = \int f(\lambda)e^{i\omega\lambda}d\lambda$で解ける。
\begin{equation}
\hat{\rho}(\omega) = \frac{\hat{a}_1(\omega)}{1 + \hat{a}_2(\omega)}
\end{equation}
これを逆変換すれば$\rho(\lambda)$と$e_0$が閉じた形で求まる。

- **有限サイズ補正**：有限の$n$では、和と積分の差から$E_0/n$に補正がつく。ギャップのない（臨界的な）系では、周期境界条件での補正は$1/n^2$に比例し、その係数は共形場理論の中心電荷$c$と励起の速度$v$で決まる（$E_0/n \approx e_0 - \frac{\pi cv}{6n^2}$）。ギャップのある系では補正は$n$について指数関数的に小さい。

## 10. 複素数の解と励起

Bethe方程式の解は実数とは限らない。

- **束縛状態とストリング**：引力的な相互作用があると、粒子どうしが束縛状態を作る。束縛状態の運動量は複素数で、波動関数は粒子間の距離について指数関数的に減衰する。$n\to\infty$では、複素数の解は$\lambda \pm i\cdot(\text{一定の間隔})$のように虚軸方向に等間隔に並ぶ組（**ストリング**）になると仮定することが多い（ストリング仮説）。
- **完全性**：実数の解と複素数の解を合わせると、各粒子数の部分空間の$\binom{n}{M}$個の固有状態がすべて得られると考えられている（細かい例外の扱いには注意が必要）。
- **素励起**：基底状態の量子数の並びにホールを作った励起は、粒子1個を取り除いたものではなく、新しい準粒子として振る舞う。スピン鎖では、これが**スピノン**（スピン1/2をもつ素励起）になる。

基底エネルギーなど基底状態の性質だけなら、7節の量子数で決まる実数の解だけで十分なことが多い。

## 11. 使える範囲と他の手法との比較

| | 厳密対角化 | Jordan-Wigner変換 | Bethe仮説 |
|---|---|---|---|
| 扱う問題 | $2^n\times2^n$行列（保存量のブロックでは$\binom{n}{M}$次元） | $2n\times2n$行列 | $M$変数の連立方程式 |
| 計算コスト | 指数関数的 | $O(n^3)$ | 1回の反復が$O(M^2)$ |
| 使える模型 | 一般のハミルトニアン | 自由フェルミオンに帰着する模型 | 可積分な模型（相互作用があってもよい） |
| 求めやすい量 | 何でも（小さい$n$） | 2次形式で書ける量 | エネルギー、熱力学量 |
| 求めにくい量 | 大きい$n$のすべて | 長いひもが残る観測量 | 相関関数、時間発展 |

- **境界条件**：上では周期境界条件で導いた。開放境界条件でも、端での反射の振幅を加えた**境界のBethe仮説**で解ける模型がある（E. K. Sklyanin, J. Phys. A 21, 2375 (1988) など）。
- **相関関数と時間発展**：固有エネルギーは簡単に求まるが、固有状態の波動関数は$M!$項の和なので、期待値や時間発展を直接計算するのは難しい（form factorなどの別の手法が必要になる）。このリポジトリでは、時間発展は厳密対角化や量子回路で計算する。
- **代数的Bethe仮説**：ここで述べた波動関数を直接書く方法（座標Bethe仮説）のほかに、転送行列とYang-Baxter方程式を使って同じ結果を代数的に導く方法（代数的Bethe仮説、量子逆散乱法）もある。
