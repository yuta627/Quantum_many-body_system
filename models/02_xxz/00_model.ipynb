# XXZ模型：今回扱う模型

このフォルダで扱う模型の定義と、記号・前提をまとめる。XXZ模型の一般的な説明は [README.md](README.md) を参照。

## 模型の定義

隣り合うスピンの間に、$xy$面内の交換相互作用と、強さの異なる$z$方向の相互作用が働く模型。

今回は以下のモデルを考える。

\begin{equation}
H = J\sum_i\left(X_iX_{i+1} + Y_iY_{i+1} + \Delta\,Z_iZ_{i+1}\right)
\end{equation}

スピン演算子$\boldsymbol{S}_i = \boldsymbol{\sigma}_i/2$で書くと$H = 4J\sum_i\left(S_i^xS_{i+1}^x + S_i^yS_{i+1}^y + \Delta S_i^zS_{i+1}^z\right)$であり、文献の式と比べるときは$J$が4倍違うことに注意する。

- **第1項・第2項（$XY$項）について**

  $\sigma^\pm = (X\pm iY)/2$を使うと
  \begin{equation}
  X_iX_{i+1} + Y_iY_{i+1} = 2\left(\sigma_i^+\sigma_{i+1}^- + \sigma_i^-\sigma_{i+1}^+\right)
  \end{equation}
  と書ける。実際に計算すると
  \begin{equation}
  \begin{aligned}
  (X_0X_1 + Y_0Y_1)\ket{01} &= 2\ket{10} \\
  (X_0X_1 + Y_0Y_1)\ket{10} &= 2\ket{01} \\
  (X_0X_1 + Y_0Y_1)\ket{00} &= (X_0X_1 + Y_0Y_1)\ket{11} = 0
  \end{aligned}
  \end{equation}
  となる。隣り合う向きの異なるスピンを入れ替える（**フリップフロップ**）項であり、$\ket{1}$を粒子とみなせば、粒子が隣のサイトへ飛び移る**ホッピング**である。$\ket{1}$の数は変わらない。

- **第3項（$ZZ$項）について**

  横磁場イジングモデルの第1項と同じイジング相互作用で、$Z_iZ_{i+1}$は隣どうしが揃っていれば$+1$、揃っていなければ$-1$である。$J\Delta>0$なら隣どうしが逆向きの方がエネルギーが低い（反強磁性的）。

$XY$項は隣どうしを「重ね合わせて」エネルギーを下げようとし、$ZZ$項は$z$方向に交互に並べようとする。この2つが競合し、その比$\Delta$で基底状態の性質が決まる。

## 前提と記号

- サイト数を$n$とし、$i = 0, 1, \dots, n-1$と番号をつける。
- 特に断らない限り**開放境界条件**（$i = n-1$と$0$の間のボンドを含めない）とする。Bethe仮説（[02_jordan_wigner.ipynb](02_jordan_wigner.ipynb)）は**周期境界条件**で使うので、そのときは明記する。
- $J>0$（$XY$項について反強磁性的）、$\Delta>-1$とする。$\Delta\le-1$は強磁性相で、基底状態が単純になる（下の副格子回転を参照）。
- 記号や量子ビットの並び順の規約は [docs/conventions.md](../../docs/conventions.md) を参照。

## 対称性

- **全磁化の保存（U(1)対称性）**：$\sum_jZ_j$は$H$と交換する。$XY$項は$\ket{01}\leftrightarrow\ket{10}$と入れ替えるだけで$\ket{1}$の数を変えず、$ZZ$項は対角だからである。そのため$H$は$\ket{1}$の個数$m = 0, 1, \dots, n$ごとのブロックに分かれる（[01_exact_diag.ipynb](01_exact_diag.ipynb)の3節）。
- **スピン反転（$\mathbb{Z}_2$対称性）**：$P = \prod_jX_j$は$H$と交換する。$P$は$Y\to-Y$、$Z\to-Z$と変えるが、$H$の各項は2個ずつの積なので変わらない。
- **$\Delta=1$でのSU(2)対称性**：$\Delta=1$では$H = J\sum_i\boldsymbol{\sigma}_i\cdot\boldsymbol{\sigma}_{i+1}$となり、$\sum_jX_j$、$\sum_jY_j$も$H$と交換する（スピンをどの方向に回転しても変わらない）。
- **副格子回転**：奇数サイトだけを$z$軸まわりに$\pi$回転させるユニタリ$U = \prod_{j:\text{奇数}}Z_j$を考える。$ZX Z = -X$、$ZYZ = -Y$なので、隣り合うサイトの一方だけで$X, Y$の符号が反転し
  \begin{equation}
  U\,H(J, \Delta)\,U^\dagger = J\sum_i\left(-X_iX_{i+1} - Y_iY_{i+1} + \Delta Z_iZ_{i+1}\right) = -H(J, -\Delta)
  \end{equation}
  となる（開放境界条件、または周期境界条件で$n$が偶数のとき）。したがって$(J, \Delta)$のスペクトルは、$(J, -\Delta)$のスペクトルの符号を反転したものになる。$\Delta<-1$の基底状態は、$-H(J,-\Delta)$の基底状態、すなわち$H(J, |\Delta|)$の**最高エネルギー状態**に対応し、これは全スピンが揃った強磁性状態である。

## いくつかの極限と相

| $\Delta$ | 模型 | 基底状態の性質 |
|---|---|---|
| $\Delta\to+\infty$ | 反強磁性イジング模型 | Néel状態 $\ket{0101\cdots}$, $\ket{1010\cdots}$ の2重縮退 |
| $\Delta>1$ | | ギャップのある反強磁性（Néel）相 |
| $\Delta=1$ | 反強磁性ハイゼンベルク模型 | 臨界的（SU(2)対称） |
| $-1<\Delta<1$ | | ギャップのない臨界相（朝永-Luttinger液体） |
| $\Delta=0$ | XX模型 | 自由フェルミオン（[02_jordan_wigner.ipynb](02_jordan_wigner.ipynb)の3節） |
| $\Delta<-1$ | | 強磁性相（全スピンが揃う） |

スタッガード磁化は
\begin{equation}
M_s = \frac{1}{n}\sum_{j}(-1)^jZ_j,\quad m_s(t) = \bra{\psi(t)}M_s\ket{\psi(t)}
\end{equation}
で定義する。初期状態のNéel状態$\ket{0101\cdots}$（偶数サイトが$\ket{0}$、奇数サイトが$\ket{1}$）では$m_s(0) = 1$である。Néel状態は$\Delta\to\infty$（イジング極限）の基底状態なので、「イジング相互作用だけの系に$XY$項を急に加えたとき（クエンチ）に、反強磁性の秩序がどう崩れていくか」を見ることになる。

TFIMの横磁化$m_x$は全体の和なので左右反転で変わらなかったが、$M_s$は$n$が偶数のとき左右反転で符号が変わる（$(-1)^{n-1-j} = -(-1)^j$）。サイトの番号の付け方をそろえて比較する必要がある（[docs/conventions.md](../../docs/conventions.md)）。
